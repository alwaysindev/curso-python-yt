## Métodos

Los métodos son **funciones especiales que pertenecen a un objeto**.  
Nos permiten modificarlo o pedirle información.  
Cada tipo de objeto (strings, listas, diccionarios, etc.) tiene sus propios métodos disponibles.

La sintaxis es:

```python
objeto.metodo()


### String
Una variable de tipo string, tiene una serie de métodos que permiten sacarle jugo a la cadena de texto. [Aquí](https://docs.python.org/2.5/lib/string-methods.html) tienes todos los métodos que podemos usar en cadenas de texto

````python
string_ejemplo = "string en mayusculas"

# Para poner un string todo en mayusculas
print("Todo mayusculas:", string_ejemplo.upper())

# Para poner un string todo en minusculas
print("Todo minusculas:", string_ejemplo.lower())

# Para sustituir caracteres.
print("Sustituir s por S:", string_ejemplo.replace("s", "S"))

# El replace también es muy útil cuando queremos eliminar caracteres.
print("Eliminar s:", string_ejemplo.replace("s", ""))

# Divide el string por un caracter en una LISTA
print("Separalo segun el numero de espacios:", string_ejemplo.split(" "))
````

### Ejercicio: métodos de strings

1. Crea una variable `frase = "  Python es Genial!!!  "`.
2. Haz lo siguiente usando métodos de strings:
   - Elimina los espacios en blanco al inicio y al final.
   - Convierte todo el texto a minúsculas.
   - Reemplaza `"genial"` por `"divertido"`.
   - Cuenta cuántas veces aparece la letra `"i"`.
3. Muestra cada resultado con `print()`.


````python
frase = "  Python es Genial!!!  "

# 1. Quitar espacios
sin_espacios = frase.strip()
print(sin_espacios)

# 2. Todo en minúsculas
minusculas = sin_espacios.lower()
print(minusculas)

# 3. Reemplazar palabra
nueva = minusculas.replace("genial", "divertido")
print(nueva)

# 4. Contar letras
conteo_i = nueva.count("i")
print("Número de 'i':", conteo_i)
````

## Métodos en Listas
En [este link](https://www.w3schools.com/python/python_ref_list.asp) puedes encontrar todos los métodos que podrás usar con listas.

````python
asignaturas = ["Mates", "Fisica", "Inglés"]
print(asignaturas)

# Para agregar nuevos elementos. No hace falta reasignarlo a la lista. Simplemente, se añade al objeto asignaturas.
asignaturas.append("Quimica")
print(asignaturas)

# Obtener el indice de un elemento de la lista
print(asignaturas.index("Fisica"))

# Eliminar todos los elementos. Fijate que devuelve None, el elemento vacio.
# asignaturas.clear()
# print(asignaturas.clear())
# print(asignaturas)
````
