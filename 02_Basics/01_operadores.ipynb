## Operadores aritméticos
- `+` suma
- `-` resta
- `*` multiplicación
- `/` división
- `//` división entera
- `%` resto
- `**` potencia


```python
print(5 + 3)   # 8
print(5 - 3)   # 2
print(5 * 3)   # 15
print(5 / 2)   # 2.5
print(5 // 2)  # 2
print(5 % 2)   # 1
print(2 ** 3)  # 8
```

```python
print("Conmutativa")
print(2 * 3)
print(3 * 2)

print("\nAsociativa")
print((3 * 2) * 5)
print(3 * (2 * 5))

print("\nDistributiva") 
print(2 * (3 + 5))
print(2 * 3 + 2 * 5)


print("\nEl Orden de operaciones se mantiene. Siempre podemos usar paréntesis")
print(2 * (2 + 3) * 5)
print((2 * 2 + 3 * 5)/(4 + 7))
```

## Operadores de comparación
- `==` igual
- `!=` distinto
- `>` mayor que
- `<` menor que
- `>=` mayor o igual
- `<=` menor o igual


```python
print(5 == 5)   # True
print(5 != 3)   # True
print(7 > 2)    # True
print(2 < 1)    # False
print(3 >= 3)   # True
print(4 <= 2)   # False
```

## Operadores lógicos
- `and` → True si **ambas** condiciones son True.
- `or` → True si **al menos una** condición es True.
- `not` → invierte el valor lógico.


```python
print(True and True)   # True
print(True and False)  # False
print(True or False)   # True
print(not False)       # True

# Con números (recordar: 0 es False, otros son True)
print(bool(5) and bool(1))   # True
print(bool(0) or bool(10))   # True
```

#### `is` e `is not`

```python
x = [5]
y = [5]

x is y
```

```python
[5] is [5]
```

1. Recordemos

== → compara valores (contenido).

is → compara identidad (¿es exactamente el mismo objeto en memoria?).

2. Caso: 5 is 5

Los números pequeños (normalmente de –5 a 256) Python los internamente “cachea” (interning).

Eso significa que siempre que escribas 5, apunta al mismo objeto en memoria.

```python
a = 5
b = 5
print(a is b)   # True → misma dirección en memoria
```

3. Caso: [5] is [5]

Cada vez que creas una lista con [5], Python construye un objeto nuevo en memoria.

Aunque tengan el mismo contenido, no son el mismo objeto.

```python
a = [5]
b = [5]
print(a == b)  # True → mismo contenido
print(a is b)  # False → objetos distintos en memoria
```

4. Analogía

Los números pequeños son como sillas numeradas en un teatro: si dices “silla 5”, todo el mundo apunta a la misma.

Las listas son como papelitos: si escribes “5” en dos papeles distintos, ambos dicen lo mismo, pero son papeles distintos.



### Ejercicio final: aplicar un descuento

1. Imagina que tienes una compra de `precio = 100`.
2. Si tienes un código de descuento igual a `"PROMO10"`, se resta 10 al precio.
3. Crea una variable `codigo = "PROMO10"` y otra `precio = 100`.
4. Calcula el nuevo precio aplicando el descuento **solo si el código es correcto** (usa un operador de comparación `==`).
5. Imprime el precio final.


# SOLUCIÓN

```python
precio = 100
codigo = "PROMO10"

# Comprobamos si el código es correcto (esto da True o False)
descuento_activo = (codigo == "PROMO10")

# Aplicamos el descuento multiplicando el booleano por 10
precio_final = precio - (10 * descuento_activo)

print("Precio final:", precio_final)
```
