## Estructuras
- **list**: mutable, ordenada → `[1,2,3]`
- **tuple**: inmutable, ordenada → `(1,2,3)`
- **set**: sin orden, sin duplicados → `{1,2,3}`
- **dict**: pares clave:valor → `{"a":1, "b":2}`


```python
lista = [1, 2, 3, 3, 3, 3]

set(lista)
```

```python
lista = [1, 2, 3]
tupla = (1, 2, 3)
conjunto = {1, 2, 2, 3}
dic = {"a": 1, "b": 2}

print(lista, type(lista))
print(tupla, type(tupla))
print(conjunto, type(conjunto))  # {1,2,3}
print(dic, type(dic))
```
