## La función `input()`
- Sirve para leer texto escrito por el usuario.
- Siempre devuelve un **string**.


```python
nombre = input("¿Cómo te llamas? ")
print("Encantado,", nombre)
```

## Convertir texto a número
- Como `input()` siempre da un string, si quieres un número debes convertirlo.
- Se usa `int()` o `float()`.


```python
edad_txt = input("¿Cuántos años tienes? ")
edad = int(edad_txt)  # convierte el string a int
print("Dentro de un año tendrás", edad + 1)
```
