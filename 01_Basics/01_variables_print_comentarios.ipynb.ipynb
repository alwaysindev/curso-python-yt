## Todo en Python es un objeto
En Python absolutamente todo es un objeto: números, textos, listas…
Podemos comprobarlo con la función `type()`.


````python
print(type("Hola"))   # string
print(type(5))        # int
print(type(3.14))     # float
print(type(True))     # bool
print(type(None))     # NoneType
````

## Strings
Un string es texto. Se escribe entre comillas.


````python
nombre = 'Alex'
````

````python
mensaje = "Hola"
print(mensaje)
print(type(mensaje))
````

## Int y Float
- Int son números enteros: 1, 2, 3…
- Float son números con decimales: 3.14, 2.5…


````python
edad = 25      # int
altura = 1.75  # float

print(edad, type(edad))
print(altura, type(altura))
````

## Booleanos
- Los booleanos vienen del álgebra de Boole.
- `True` equivale a 1, `False` equivale a 0.


````python
es_mayor = True
print(es_mayor, type(es_mayor))

print(True + True)   # 2
print(True + False)  # 1
````

## Listas
Una lista es un objeto que guarda otros objetos de forma secuencial.
Lo veremos más en detalle en otra clase.


````python
numeros = [1, 10, 3,4,7,8, None, 5]
print(numeros, type(numeros))
````

## Variables
Cuando escribimos código, guardamos objetos en variables.
- Las variables ayudan a dar contexto: qué representa ese valor.
- Por convención, van en minúsculas.
- Si quieres un valor fijo (constante), se escribe en mayúsculas.


````python
edad = 25
API_KEY = "12345"

print(edad)
print(API_KEY)
````

## Funciones
- Una función es un bloque de código que hace una tarea específica.
- Python trae funciones ya hechas: son las *builtin functions*.
- La más usada es `print()`, sirve para mostrar algo en pantalla.


````python
print("Hola mundo")
````

## Comentarios
Los comentarios sirven para añadir notas en el código.
Python ignora los comentarios.


````python
# Esto es un comentario

"""
Esto es un comentario
que ocupa varias líneas
"""

"Hola"
````

En Python, \n significa salto de línea. Es un carácter especial dentro de un string que indica: “aquí baja a la siguiente línea”.
