## Conversión de tipos
- `int("10")` → `10`, pero `int("10.5")` **falla**.
- `float("10.5")` → `10.5`.
- `str(123)` → `"123"`.
- `bool(x)` → `False` para vacíos/0/None; `True` en el resto.


```python
print(int("10"), type(int("10")))
print(float("10.5"), type(float("10.5")))
print(str(123), type(str(123)))

# int("10.5")  
```
