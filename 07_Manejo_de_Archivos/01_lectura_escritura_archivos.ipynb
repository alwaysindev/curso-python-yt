## 📄 1. Trabajando con archivos de texto (`.txt`)


````python
# Escribir un archivo de texto con write() — recibe un string
texto = "Hola mundo\nEsta es una segunda línea\nY otra más\n"

with open("archivo.txt", "w", encoding="utf-8") as f:
    f.write(texto)

print("Archivo escrito con write().")
````

````python
# write() solo acepta strings — esto da TypeError:
lista = ["Hola mundo\n", "Esta es una segunda línea\n", "Y otra más\n"]

try:
    with open("archivo.txt", "w", encoding="utf-8") as f:
        f.write(lista)   # ❌ TypeError: write() argument must be str, not list
except TypeError as e:
    print("Error:", e)
    print("→ Usa writelines() para escribir una lista, o únela con ''.join(lista)")
````

````python
# writelines() sí acepta una lista de strings
lista = ["Hola mundo\n", "Esta es una segunda línea\n", "Y otra más\n"]

with open("archivo.txt", "w", encoding="utf-8") as f:
    f.writelines(lista)   # ✅ escribe cada elemento seguido

print("Archivo escrito con writelines().")
````

````python
with open("archivo.txt", "w") as f:
    for i in lista:
        f.write(i)
````

````python
# -------------------------------------------------------------------
# Leer un archivo de texto
# -------------------------------------------------------------------

with open("archivo.txt", "r") as f:
    contenido = f.read()

print(contenido)
````

````python
# Modos de apertura:
#   "w"  → escritura (sobreescribe si existe)
#   "a"  → append (añade al final sin borrar)
#   "r"  → lectura (por defecto)
#   "rb" → lectura binaria

# Añadir una línea sin borrar el contenido anterior
with open("archivo.txt", "a", encoding="utf-8") as f:
    f.write("Línea añadida con modo 'a'\n")

with open("archivo.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

````python
# -------------------------------------------------------------------
# Leer archivo línea por línea
# -------------------------------------------------------------------

with open("archivo.txt", "r") as f:
    for linea in f:
        print("Línea:", linea.strip())  # strip para quitar saltos de línea extra
````

````python
# readlines() devuelve una lista, una línea por elemento
with open("archivo.txt", "r", encoding="utf-8") as f:
    lineas = f.readlines()

print(lineas)
print("Número de líneas:", len(lineas))
````

## 📑 2. Archivos CSV

Un archivo CSV (Comma Separated Values) es una tabla donde cada línea representa una fila, y los valores están separados por comas.

Python incluye el módulo `csv` para trabajar con este tipo de archivos sin instalar nada.

---


````python
# -------------------------------------------------------------------
# Escribir un archivo CSV
# -------------------------------------------------------------------
import csv

datos = [
    ["Nombre", "Edad", "Ciudad"],
    ["Ana", 25, "Lima"],
    ["Luis", 30, "Bogotá"],
    ["María", 22, "Buenos Aires"]
]

with open("personas.csv", "w", newline="", encoding='utf-8') as f:
    escritor = csv.writer(f)
    escritor.writerows(datos)
````

````python
# Leer un archivo CSV — usar el mismo encoding con el que se escribió
with open("personas.csv", "r", encoding="utf-8") as f:
    lector = csv.reader(f)
    for fila in lector:
        print(fila)
````

## 🧩 3. Archivos JSON

Los archivos JSON (`.json`) son muy usados para guardar datos estructurados (como diccionarios o listas de diccionarios).

Python trae un módulo llamado `json` para codificar y decodificar este tipo de archivos.

---


````python
# -------------------------------------------------------------------
# Escribir un archivo JSON
# -------------------------------------------------------------------
import json

persona = {
    "nombre": "Ana",
    "edad": 25,
    "intereses": ["lectura", "programación"]
}

with open("persona.json", "w") as f:
    json.dump(persona, f, indent=4)
````

````python
personas = [
    {"nombre": "Ana", "edad": 25, "intereses": ["lectura", "programación"]},
    {"nombre": "Luis", "edad": 30, "intereses": ["deportes", "música"]},
    {"nombre": "María", "edad": 22, "intereses": ["arte", "cine"]}
]

with open("personas.json", "w") as f:
    json.dump(personas, f, indent=4)  # Escribir la lista de personas en el archivo JSON

    '''el parámetro indent controla la indentación (sangría) del JSON generado,
    es decir, cuántos espacios se usan para anidar niveles.'''
````

````python
# -------------------------------------------------------------------
# Leer un archivo JSON
# -------------------------------------------------------------------

with open("persona.json", "r") as f:
    datos = json.load(f)

print("Nombre:", datos["nombre"])
print("Intereses:", datos["intereses"])
````

````python
with open("personas.json", "r") as f:
    datos = json.load(f)
print(datos)
print(datos[0])
print(datos[0]['intereses'])
print(datos[0]['intereses'][0])

print('-----------------------------------------------------')
for persona in datos:
    print("Nombre:", persona["nombre"])
    print("Edad:", persona["edad"])
    print("Intereses:", persona["intereses"])
    print()
````

````python
import json, os

def registrar_usuario(nombre, edad):
    ruta = "usuarios.json"
    if os.path.exists(ruta):
        with open(ruta, "r", encoding="utf-8") as f:
            usuarios = json.load(f)
    else:
        usuarios = []

    usuarios.append({"nombre": nombre, "edad": edad})

    with open(ruta, "w", encoding="utf-8") as f:
        json.dump(usuarios, f, indent=4, ensure_ascii=False)
    print(f"Usuario '{nombre}' registrado. Total: {len(usuarios)}")

registrar_usuario("Ana", 28)
registrar_usuario("Luis", 34)
registrar_usuario("Marta", 22)

with open("usuarios.json", "r", encoding="utf-8") as f:
    print(json.load(f))
````

Crea una función `registrar_usuario(nombre, edad)` que:
1. Lea un archivo `usuarios.json` si existe (o empiece con lista vacía).
2. Añada el nuevo usuario `{"nombre": ..., "edad": ...}`.
3. Guarde la lista actualizada en `usuarios.json`.

````python
conteo = {}

with open("archivo.txt", "r", encoding="utf-8") as f:
    for linea in f:
        for palabra in linea.lower().split():
            palabra = palabra.strip(".,;:!?")
            conteo[palabra] = conteo.get(palabra, 0) + 1

for palabra, freq in sorted(conteo.items(), key=lambda x: x[1], reverse=True):
    print(f"{palabra}: {freq}")
````
